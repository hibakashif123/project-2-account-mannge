```python
def get_positive_number(prompt):
    while True:
        try:
            amount = float(input(prompt))
            if amount <= 0:
                print("invalid input: that is not a positive!")
                continue
            return amount
        except ValueError:
          print("That is invalid!")
name = input(" what is your name?")
Balance = get_positive_number(" what is your starting balance? $")
while True:
  account_type = input(" what account you want?(savings or checkings)")
  if account_type in [ "savings", "checkings"]:
     break
  else:
    print(" it is not in the options!")

print(f"\n Account Created! welcome {name}")
print(f" balnce: ${Balance}, account_type: {account_type}")
while True:
    print("==Menu==")
    print("1.Check Balance")
    print("2.Deposit")
    print("3. Withdraw")
    print("4.Quit")

    try:
      choose =int(input("choose from 1-4"))
      if choose < 1 or choose > 4:
       print("That is invalid!")
       continue
    except ValueError:
      print("Please enter a number!")
      continue
    if choose ==1:
      print(f"\n check balance: ${Balance}\n")
    elif choose == 2:
        Deposit = get_positive_number("what is the deposit? $")
        Balance += Deposit
        print(f"Deposited: ${Deposit:.2f}. New Balance: ${Balance:.2f}\n")
    elif choose == 3:
        Withdraw = get_positive_number("what is the withdraw? $")
        Balance -= Withdraw
        print(f"Withdraw: ${Withdraw:.2f}. New Balance: ${Balance:.2f}\n")
    else:
        print(f"Thank you, {name} for working with us")
        break








```
